```python
import numpy as np
import matplotlib.pyplot as plt

N = 1000
fs = 500


def mi_funcion_sen(vmax=1.5, dc=0, ff=3, ph=0, nn=N, fs=fs):
    tt = (np.arange(nn) / fs).reshape(nn, 1)
    xx = dc + vmax * np.sin(2 * np.pi * ff * tt + ph)
    return tt, xx

tt, xx = mi_funcion_sen(vmax=1.5, dc=0, ff=3, ph=np.pi/2, nn=N, fs=fs)

plt.figure(figsize=(10, 4))
plt.plot(tt, xx, linewidth=1.5, color='steelblue')
plt.title('Señal senoidal generada')
plt.xlabel('Tiempo [s]')
plt.ylabel('Amplitud [V]')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# la funcion de plot interpola
```
