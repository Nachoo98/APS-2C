```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 43)

casos = [
    (np.pi / 6, "w0 = pi / 6"),
    (3.0, "w0 = 3"),
    (4 * np.pi / 7, "w0 = 4pi / 7")
]

fig, axs = plt.subplots(3, 1, figsize=(9, 7), sharex=True)

for ax, (w0, titulo) in zip(axs, casos):
    ax.stem(n, np.cos(w0 * n))
    ax.set_title(titulo)
    ax.grid(True)

axs[-1].set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib . pyplot as plt

rng = np . random . default_rng (0)
patron = np . array ([2.0 , 3.0 , 1.0])
D = 40 # retardo a detectar
N = 100
y = 2.0 * rng . standard_normal ( N ) # ruido de fondo
y [ D : D +3] += patron # eco escondido en el ruido
# correlacion cruzada : se desliza el patron sobre y, sin dar vuelta nada
r = np . correlate (y , patron , mode ="full")
lags = np . arange ( - len( patron ) + 1 , N )
fig , ( a1 , a2 ) = plt . subplots (2 , 1 , figsize =(9 , 5) )
a1 . plot ( y ) ; a1 . set_title (" Senal recibida (el eco no se ve)")
a2 . plot ( lags , r ); a2 . set_title (" Correlacion cruzada r_yx ")
a2 . set_xlabel (" retardo l")
for a in ( a1 , a2 ) : a . grid ( True )
plt . tight_layout ()
plt . show ()
print (" Pico en l =", lags [ np . argmax ( r ) ])
print (" Valor del pico =", r [ np . argmax ( r ) ] , " ( comparar con E_patron = 14)")
```
